```python
df = pd.read_csv('wine-clustering.csv', usecols=[0,1,2])
df.columns=['Alcohol', 'Malic acid', 'Ash']
```

```python
df
```

```python
# Convert string values to numeric, handling any non-numeric values
df['Alcohol_numeric'] = pd.to_numeric(df['Alcohol'], errors='coerce')

# Remove the non-numeric value ('Malic_Acid') and any NaN values
df_clean = df.dropna(subset=['Alcohol_numeric'])

# Now create the kde plot with the cleaned numeric data
sns.kdeplot(df_clean['Alcohol_numeric'])
plt.title('KDE Plot of Alcohol Content')
plt.show()

# Optional: Check how many values were removed
print(f"Original rows: {len(df)}")
print(f"Rows after cleaning: {len(df_clean)}")
print(f"Removed {len(df) - len(df_clean)} non-numeric values")
```

#1.) Converted string values to numeric
2.) errors = coerce , if a value cant be converted to a number , Pandas replaces it with Nan (not a number) instaed of giving an error
3.) df_clean = df,dropna, removes rows containing missing values
subset=['Alcohol_numeric'], Pandas checks only Alcohol_numeric column
4.) draw kde plot
5.) display following parameters
